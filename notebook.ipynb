```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from openai import OpenAI
openai_client = OpenAI()
```

```python
def llm(prompt):
    response = openai_client.responses.create(
        model='gpt-5.4-nano',
        input=prompt,
        max_output_tokens=16
    )
    return response.output_text
```

```python
question = 'what the best world cup team right now?'
answer = llm(question)
print(f'Q: {question}')
```
